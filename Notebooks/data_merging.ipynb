# 1. Import Data & Preview

```python
import numpy as np
import pandas as pd
pd.set_option('display.max_column', None)

import warnings
warnings.filterwarnings('ignore')
```

```python
demo = pd.read_excel('../Datasets/Raw/Telco_customer_churn_demographics.xlsx')
loc = pd.read_excel('../Datasets/Raw/Telco_customer_churn_location.xlsx')
pop = pd.read_excel('../Datasets/Raw/Telco_customer_churn_population.xlsx')
serv = pd.read_excel('../Datasets/Raw/Telco_customer_churn_services.xlsx')
stat = pd.read_excel('../Datasets/Raw/Telco_customer_churn_status.xlsx')

dataframes = [demo, loc, pop, serv, stat]
```

```python
for dataframe in dataframes:
    display(dataframe.head())
```

# 2. Merge Data

```python
demo = demo.drop(['Count', 'Dependents'], axis=1)
loc = loc.drop(['Count', 'Lat Long'], axis=1)
pop = pop.drop(['ID'], axis=1)
serv = serv.drop(['Count', 'Quarter', 'Referred a Friend', 'Offer'], axis=1)
stat = stat.loc[:, ['Customer ID', 'Churn Label']]
```

```python
df_master = demo.merge(loc, on='Customer ID', how='inner')
df_master = df_master.merge(serv, on='Customer ID', how='inner')
df_master = df_master.merge(stat, on='Customer ID', how='inner')
df_master = df_master.merge(pop, on='Zip Code', how='inner')
```
